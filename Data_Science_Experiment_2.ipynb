Topic 7: Functions & Passing Arguments

```python
# Q7.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-5))
print(my_abs(0))
```

```python
# Q 7.2
print(my_abs("Hi")) #This produces a TypeError because Python cannot compare a string with an integer using <.
```

```python
# Q 7.3
def print_abs(val):
    if val < 0:
        print(0 - val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)
```

print_abs() prints the value but does not return anything. Therefore, the function returns None by default.

```python
# 7.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)

print(x)
```

val is a local variable inside the function. Reassigning it does not change the original variable x.

```python
# 7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print("Inside function:", val1, val2)

x = 6
y = 3

swap_message(x, y)

print("Original variables:", x, y)
```

The original variables remain unchanged because the function only swaps its local variables.

Topic 8: Scope — Local vs Global Variables

```python
# 8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(val)
```

This produces a NameError because val is a local variable that exists only inside the function.

```python
# 8.2
counter = 0

def show_counter():
    print(counter)

show_counter() # A function can read a global variable
```

```python
# 8.3
counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()
```

This produces an UnboundLocalError because Python treats counter as a local variable inside the function due to the assignment.

```python
# 8.4
counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
