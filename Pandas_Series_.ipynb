```python
#  Q29. Complete Pandas Program
import pandas as pd

# Create a labelled Series
sales = pd.Series([450,700,550,300],index =['Pen','Bag','Notebook','Bottle'])

# Display values
print("Values:")
print(sales.values)

# Display index
print("\nIndex:")
print(sales.index)

# Select one label
print("\nSales of Bag:")
print(sales['Bag'])

# Filter sales above 500
print("\nSales above 500:")
print(sales[sales > 500])

# Check whether a label exists
print("\nIs Pen present?")
print('Pen' in sales.index)

# Convert Series to dictionary
print("\nDictionary:")
print(sales.to_dict())
```

```python
# Q30. Code Analysis
data = {'A': 15, 'B': 35, 'C': 25}

s = pd.Series(data, index=['C', 'A', 'D', 'B'])

s.name = 'Observations'

s.index.name = 'Category'

print(s)
```

(a) Index order → C, A, D, B

(b) NaN → D

(c) s[s > 20] →
    C    25
    B    35

(d) pd.notnull(s) →
    C → True
    A → True
    D → False
    B → True

(e) Series name → Observations
    Index name  → Category
